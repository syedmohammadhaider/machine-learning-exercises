# Support Vector Machines (SVM)

- It is a powerful machine learning model, capable of performing linear and non-linear classification, regression and outlier detection.
- It is well suited for classification of complex small or medium sized datasets.

## Linear SVM Classification
- SVM classifiers fit the widest possible "street" between different classes.
- The boundary is determined by the instances located on the edge of the "street" (called _support vectors_), and hence remain unchanged if more training instances are added "off the street".
- SVMs are sensitive to feature scales.

### Soft Margin Classification
- **Hard margin classficiation**: all instances stay on "their side" of the decision boundary.
- Problems associated with hard margin classification:
  1. only works with linearly separable data (data that can be separated using a line)
  2. it is sensitive to outliers
- **Soft margin classification** is used to avoid these issues.
- The objective is to find a good balance between keeping the street as large as possible and limiting the margin violations (instances on or over the other side of the street).
- When we create an SVM model in Scikit-Learn, we can specify a number of hyperparameters.
- We can tune `C` to a low value if we want to ignore large number of margin violations, and vice versa.
- An overfitting SVM model should be regularized by reducing `C`. 

```python
import numpy as np
from sklearn import datasets
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
```

```python
iris = datasets.load_iris()
X = iris["data"][:, (2, 3)] # petal length, petal width
y = (iris["target"] == 2).astype(np.float64)

svm_clf = Pipeline([
    ("scaler", StandardScaler()), 
    ("linear_svc", LinearSVC(C=1, loss="hinge"))
])

svm_clf.fit(X, y)
```

```python
svm_clf.predict([[5.5, 1.7]])
```

**Note**: Instead of using LinearSVC we could use SVC class with `kernel` set to `"linear"` (`SVC(kernel="linear", C=1`) or we could use `SGDClassifier(loss="hinge", alpha=1/(m*C))`.

## Nonlinear SVM Classification

- Many datasets aren't linearly separable.
- One way this can be handled is by is to add more features such as polynomial features. 

```python
from sklearn.datasets import make_moons
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

X, y = make_moons(n_samples=100, noise=0.15)
polynomial_svm_clf = Pipeline([
    ("poly_features", PolynomialFeatures(degree=3)),
    ("scaler", StandardScaler()), 
    ("svm_clf", LinearSVC(C=10, loss="hinge"))
])

polynomial_svm_clf.fit(X, y)
```

- Another way to handle nonlinear data is called the _kernel trick_.
- It makes it possible to get same result as if we had added many polynomial features even with high degrees, without actually having to add them. 

```python
from sklearn.svm import SVC

poly_kernel_svm_clf = Pipeline([
    ("scaler", StandardScaler()), 
    ("svm_clf", SVC(kernel="poly", degree=3, coef0=1, C=5)) # uses 3rd degree polynomial kernel
])
poly_kernel_svm_clf.fit(X, y)
```

### Similarity Features

- Another technique to handle nonlinear problems is to add features computed using a similarity function.
- It measures how much each instance resembles a particular landmark.

**Gaussian RBF**: $$\phi_{\gamma}(\textbf{x}, l) = \text{exp}(-\gamma \lVert \textbf{x} - l \rVert ^2) $$

```python
rbf_kernel_svm_clf = Pipeline([
    ('scaler', StandardScaler()),
    ('svm_clf', SVC(kernel="rbf", gamma=5, C=0.001))
])
rbf_kernel_svm_clf.fit(X, y)
```
